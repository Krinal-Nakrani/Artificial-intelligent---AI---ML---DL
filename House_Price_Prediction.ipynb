```python
House Price Prediction (Regression)

Problem: Predict the price of a house given features like size, location, rooms, etc.
Input: Numerical features (sq ft, location encoded, etc.).
Output: Continuous value (house price).
Why: Real-world business use case, introduces regression with neural networks.

What to learn: 
Difference between classification and regression in TensorFlow
Using mean_squared_error loss
Scaling input features for better accuracy
```

```python
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
data= pd.read_csv("Housing.csv")
data = data.dropna() #handle missing data values
```

```python
#Conver categorical colun
data = pd.get_dummies(data, columns = ["ocean_proximity"], drop_first = True)
```

```python
#features and targets
X = data.drop("median_house_value", axis = 1)
y = data["median_house_value"]
```

```python
#train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
#scaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = Sequential([
    Dense(64, activation = 'relu', input_shape = (X_train.shape[1],)),
    Dense(32, activation = 'relu'),
    Dense(1)
])
```

```python
#compile
model.compile(optimizer = 'adam', loss = 'mse', metrics = ['mae'])
```

```python
#train
model.fit(X_train, y_train, epochs = 200, batch_size = 32, validation_split = 0.2)
```

```python
loss, mae = model.evaluate(X_test, y_test)
```

```python
print(f"test MAE (Mean Absolute Error): {mae:.2f}")
```
